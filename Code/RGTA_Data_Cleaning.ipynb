## STEP 2 DATA CLEANING

```python
df.isnull().sum()
```

```python
df.isnull().sum() / len(df)*100 
```

```python
df['Region'] = df['Region'].fillna(df['Region'].mode()[0])

df['Plan_Type'] = df['Plan_Type'].fillna(df['Plan_Type'].mode()[0])

df['Acquisition_Channel'] = df['Acquisition_Channel'].fillna(
    df['Acquisition_Channel'].mode()[0]
)
```

```python
df['Seats'] = df['Seats'].fillna(df['Seats'].median())

df['MRR_Amount'] = df['MRR_Amount'].fillna(df['MRR_Amount'].median())
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.dtypes
```

```python
df["Month_Number"] = pd.to_numeric(df["Month_Number"], errors="coerce")
df["Seats"] = pd.to_numeric(df["Seats"], errors="coerce")
df["MRR_Amount"] = pd.to_numeric(df["MRR_Amount"], errors="coerce")
df["Total_Company_MRR"] = pd.to_numeric(df["Total_Company_MRR"], errors="coerce")
df["Active_Customer_Count"] = pd.to_numeric(df["Active_Customer_Count"], errors="coerce")
```

```python
df.dtypes
```

```python
df["Region"] = df["Region"].str.strip()
df["Industry"] = df["Industry"].str.strip()
df["Plan_Type"] = df["Plan_Type"].str.strip()
df["Acquisition_Channel"] = df["Acquisition_Channel"].str.strip()
df["Customer_Status"] = df["Customer_Status"].str.strip()
df["Phase"] = df["Phase"].str.strip()
```

```python
df["Region"].unique()
```

```python
df["Plan_Type"].unique()
```

```python
df["Customer_Status"].unique()
```

```python
df["Region"] = df["Region"].str.title()
df["Plan_Type"] = df["Plan_Type"].str.title()
df["Customer_Status"] = df["Customer_Status"].str.title()
```

```python
df["Region"].unique()
```

```python
(df["Seats"] < 0).sum()
```

```python
(df["MRR_Amount"] < 0).sum()
```

```python
(df["Total_Company_MRR"] < 0).sum()
```

```python
df.loc[df["Seats"] < 0, "Seats"] = np.nan
df.loc[df["MRR_Amount"] < 0, "MRR_Amount"] = np.nan
df.loc[df["Total_Company_MRR"] < 0, "Total_Company_MRR"] = np.nan
```

```python
df.isnull().sum()
```

```python
df["Seats"] = df["Seats"].fillna(df["Seats"].median())

df["MRR_Amount"] = df["MRR_Amount"].fillna(df["MRR_Amount"].median())

df["Total_Company_MRR"] = df["Total_Company_MRR"].fillna(
    df["Total_Company_MRR"].median()
)
```

```python
df.isnull().sum()
##df[df["Month_Date"].isna()]
```

```python
df["Month_Date"] = pd.to_datetime(df["Month_Date"], errors="coerce")

df["Month_Date"] = df.groupby("Month_Number")["Month_Date"].transform("first")
```

```python
df["Month_Date"].isnull().sum()
```

```python
# Check actual duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```
